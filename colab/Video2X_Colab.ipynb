```python
import os, shutil, urllib.request, zipfile
REPO_URL = "https://github.com/shaileshdeshpande55/upscale-video-4k-collab-video2x"
ZIP_URL = REPO_URL + "/archive/refs/heads/main.zip"
DEST = "/content/app"
if os.path.exists(DEST):
    shutil.rmtree(DEST)
zip_path = "/content/repo.zip"
urllib.request.urlretrieve(ZIP_URL, zip_path)
with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall("/content")
extracted = "/content/upscale-video-4k-collab-video2x-main"
if os.path.exists(extracted):
    os.rename(extracted, DEST)
print("Downloaded. Files:", os.listdir(DEST + "/colab"))
```
